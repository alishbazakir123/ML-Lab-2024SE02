## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).
- **(g)** Monsoon (Jul–Sep) rain of each city by slicing columns.
- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# T7: your code here
# T7a: Shape of tmax
print("T7a (Shape):", tmax.shape)

# T7b: Annual mean max temp per city
t7b = tmax.mean(axis=1)
print("T7b (Annual Mean Tmax °C):", t7b)

# T7c: Hottest month of each city
t7c = months[tmax.argmax(axis=1)]
print("T7c (Hottest Month):", t7c)

# T7d: 5-city monthly average
t7d = tmax.mean(axis=0)
print("T7d (5-City Monthly Avg °C):", t7d)

# T7e: Months > 35 °C
t7e = (tmax > 35).sum(axis=1)
print("T7e (Months > 35°C):", t7e)

# T7f: Lahore temp in °F
t7f = tmax[0] * 9/5 + 32
print("T7f (Lahore °F):", t7f)

# T7g: Monsoon rain (Jul-Sep) per city
t7g = rain[:, 6:9].sum(axis=1)
print("T7g (Monsoon Rain mm):", t7g)

# T7h: Wettest monsoon city & monsoon share %
wettest_monsoon = cities[t7g.argmax()]
monsoon_pct = (t7g / rain.sum(axis=1)) * 100
print("T7h (Wettest Monsoon City):", wettest_monsoon)
print("T7h (Monsoon Share %):", monsoon_pct)
```

# Interpret
Surprising Result: It is surprising to find that Islamabad receives significantly higher monsoon rainfall (July–September) than Lahore (~763 mm vs. ~454 mm), despite Lahore having higher overall mean maximum temperatures.


## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
`cpi` has 2 rows (FY2024, FY2025) and 10 columns (Jul to Apr), year-on-year CPI inflation in %.
- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.
- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# T8a: Average inflation of each year (FY2024, FY2025)
t8a = cpi.mean(axis=1)
print("T8a: Average inflation FY2024 vs FY2025 (%):", t8a)

# T8b: FY2024 months with inflation above 25 %
t8b = cpi_months[cpi[0] > 25]
print("T8b: FY2024 months above 25 %:", t8b)

# T8c: Drop in inflation for each month (FY2024 - FY2025), in % points
t8c = cpi[0] - cpi[1]
print("T8c: Drop in inflation per month (% points):", t8c)

# T8d: Month with the biggest drop
t8d = cpi_months[t8c.argmax()]
print("T8d: Month with biggest drop:", t8d, "| drop =", t8c.max(), "% points")

# T8e: Label each FY2025 month: Low (< 5), Moderate (5-10), High (>= 10)
t8e = np.where(cpi[1] < 5, "Low", np.where(cpi[1] < 10, "Moderate", "High"))
print("T8e: FY2025 labels:", t8e)
```

# Interpret
Price Trends Between the Two Years: Pakistan's CPI inflation declined sharply between FY2024 and FY2025, falling from severe double-digit peaks (averaging ~26.2%) down to much lower single digits (averaging ~4.9%), indicating that the upward pressure on prices cooled down drastically.

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9: your code here
# T9a: Add 5 grace marks capped at 100
t9a = np.clip(marks + 5, 0, 100)
print("T9a (Grace Marks):", t9a)

# T9b: Pass % after grace
t9b = (t9a >= 50).mean() * 100
print("T9b (Pass %):", t9b)

# T9c: Indices of students who passed ONLY due to grace
t9c = np.where((marks < 50) & (t9a >= 50))[0]
print("T9c (Passed due to Grace Indices):", t9c)

# T9d: Z-score of each student after grace
t9d = (t9a - t9a.mean()) / t9a.std()
print("T9d (Z-scores):\n", t9d)
```

## Task 10 · An image is an array
- **(a)** Start from `np.zeros((8, 8), dtype=int)` and draw the "7" from the lecture with slicing (white = 255, grey = 128).
- **(b)** Flip it left-right.
- **(c)** Invert it (`255 - img`).
- **(d)** Crop rows 1–4 and columns 3–6.
- **(e)** Brighten by 80, capped at 255.
- **(f)** Print its shape, dtype and number of white pixels.

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`

```python
import numpy as np
import matplotlib.pyplot as plt

# T10a: draw the "7" on an 8x8 canvas (white = 255, grey = 128)
img = np.zeros((8, 8), dtype=int)
img[1, 1:7] = 255      # top bar (row 1, columns 1 to 6)
img[2:7, 5] = 255      # vertical stem (rows 2 to 6, column 5)
img[4, 3:5] = 128      # grey crossbar, to the left of the stem (does not overwrite it)
print("T10a: Original 7:\n", img)

# T10b: flip left-right
flipped = img[:, ::-1]
print("T10b: Flipped left-right:\n", flipped)

# T10c: invert
inverted = 255 - img
print("T10c: Inverted:\n", inverted)

# T10d: crop rows 1-4 and columns 3-6
cropped = img[1:5, 3:7]
print("T10d: Cropped (rows 1-4, cols 3-6):\n", cropped)

# T10e: brighten by 80, capped at 255
bright = np.clip(img + 80, 0, 255)
print("T10e: Brightened +80 (capped at 255):\n", bright)

# T10f: shape, dtype and number of white pixels
print("T10f: Shape:", img.shape)
print("T10f: Dtype:", img.dtype)
print("T10f: White pixels (255):", (img == 255).sum())

# Optional: show all versions side by side
titles = ["Original", "Flipped", "Inverted", "Cropped", "Brightened"]
images = [img, flipped, inverted, cropped, bright]
plt.figure(figsize=(14, 3))
for i, (t, im) in enumerate(zip(titles, images), start=1):
    plt.subplot(1, 5, i)
    plt.title(t)
    plt.imshow(im, cmap="gray", vmin=0, vmax=255)
    plt.axis("off")
plt.tight_layout()
plt.show()
```
