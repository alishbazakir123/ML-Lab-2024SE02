# Task 07 
 Function: grab the domain from "user@domain.com"

```python
def domainGet(email):
    return email.split('@')[-1]

# Example Usage:
domainGet('user@domain.com')
```

# Task 08 
Function: return True if the word "dog" is in a string (case-insensitive)

```python
def findDog(st):
    return 'dog' in st.lower()

# Example Usage:
findDog('Is there a Dog here?')
```

# Task 09
Function: count how many times "dog" occurs in a string (case-insensitive)

```python
def countDog(st):
    return st.lower().count('dog')

# Example Usage:
countDog('This dog runs faster than the other dog!')
```

# Task 10 
caught_speeding(speed, is_birthday):$\le 60 \rightarrow$ "No Ticket", $61\text{--}80 \rightarrow$ "Small Ticket", $\ge 81 \rightarrow$ "Big Ticket".If is_birthday is True, speed limits are $5\text{ mph}$ higher in all categories. 

```python
def caught_speeding(speed, is_birthday):
    # Adjust speed threshold if it's the driver's birthday
    if is_birthday:
        speed = speed - 5  # Effectively adds 5 mph allowance
        
    # Evaluate speed limits
    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"
        # Test Case 1: Speed is 60, not a birthday
print(caught_speeding(60, False)) 
# Test Case 2: Speed is 65, on birthday
print(caught_speeding(65, False))  
# Test Case 2: Speed is 65, on birthday
print(caught_speeding(65, True))  
# Test Case 3: Speed is 81, on birthday
print(caught_speeding(81, False))  
# Test Case 4: Speed is 85, not a birthday
print(caught_speeding(85, True))
```
